Step 1: Import the libraries

```python
import cv2
import matplotlib.pyplot as plt
```

Step 2: Load and display the image

```python
img = cv2.imread('Bird.png')
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')
plt.show()
```

Step 3: Convert the image into grayscale

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(gray, cmap='gray')
plt.title('Grayscale Image')
plt.axis('off')
plt.show()
```

Step 4: Apply Gaussian Blur

```python
blur = cv2.GaussianBlur(gray, (5, 5), 0)
plt.imshow(blur, cmap='gray')
plt.title('Gaussian Blur Image')
plt.axis('off')
plt.show()
```

Step 5: Apply Canny Edge Detection

```python
edges = cv2.Canny(blur, 100, 200)
plt.imshow(edges, cmap='gray')
plt.title('Canny Edge Detection')
plt.axis('off')
plt.show()
```

Step 6: Display all images together

```python
plt.figure(figsize=(12, 5))
plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')
plt.subplot(1, 3, 2)
plt.imshow(gray, cmap='gray')
plt.title('Grayscale Image')
plt.axis('off')
plt.subplot(1, 3, 3)
plt.imshow(edges, cmap='gray')
plt.title('Canny Edge Detection')
plt.axis('off')
plt.show()
```

Step 7: Change threshold values

```python
edges1 = cv2.Canny(blur, 30, 100)
edges2 = cv2.Canny(blur, 50, 150)
plt.figure(figsize=(12, 8))
plt.subplot(2, 2, 1)
plt.imshow(edges1, cmap='gray')
plt.title('Threshold 30, 100')
plt.axis('off')
plt.subplot(2, 2, 2)
plt.imshow(edges2, cmap='gray')
plt.title('Threshold 50, 150')
plt.axis('off')
```
